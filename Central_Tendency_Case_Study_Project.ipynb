**Task 2 - View the distributions of the various features in the dataset and calculate their central tendencies**

```python
# Create a histogram f the "Fixed acidity" feature

plt.figure(figsize = (11, 6))

sns.histplot(data = data ,x = 'fixed acidity', color = 'orange', edgecolor = 'linen', alpha = 0.5,
             bins = 5)
plt.title("Histogram of Fixed Acidity")
plt.xlabel('Fixed Acidity')
plt.ylabel('Count')
plt.show()
```

```python
# Calculate the mean of "fixed acidity" feature

print(round(data['fixed acidity'].mean(),2))
```

```python
# Calculate the median of "fixed acidity" feature

data['fixed acidity'].median()
```

```python
# Create a histogram of the "fixed acidity" feature and also show the mean and the median

plt.figure(figsize = (11, 6))

sns.histplot(data = data ,x = 'fixed acidity', color = 'orange', edgecolor = 'linen', alpha = 0.5,
             bins = 5)
plt.title("Histogram of Fixed Acidity")
plt.xlabel('Fixed Acidity')
plt.ylabel('Count')
plt.vlines(data['fixed acidity'].mean(), ymin = 0, ymax = 4000, color = 'blue', label = 'Mean' )
plt.vlines(data['fixed acidity'].median(), ymin = 0, ymax = 4000, color = 'red', label = 'Median' )
plt.legend()
plt.show()
```

```python
# Create a histogram of the "volatile acidity" feature

plt.figure(figsize = (11, 6))

sns.histplot(data = data ,x = 'volatile acidity', color = 'green', edgecolor = 'linen', alpha = 0.5,
             bins = 5)
plt.title("Histogram of Volatile Acidity")
plt.xlabel('Volatile Acidity')
plt.ylabel('Count')
plt.show()
```

```python
# Plot distplot using "volatile acidity" feature

plt.figure(figsize = (11, 6))

# sns.distplot(data['volatile acidity'], color = 'blue')
sns.displot(data['volatile acidity'], color = 'blue')     # `distplot` is a deprecated function and will be removed in seaborn v0.14.0.
plt.title("Distplot of Volatile Acidity")
plt.xlabel('Volatile Acidity')
plt.ylabel('Density')
plt.show()
```

```python
# Calculate the mean of "Volatile Acidity" feature

print(data['volatile acidity'].mean())
```

```python
# Create a histogram of the "citric acid" feature
plt.figure(figsize = (11, 6))

sns.histplot(data = data ,x = 'citric acid', color = 'red', edgecolor = 'linen', alpha = 0.5,
             bins = 5)
plt.title("Histogram of Citric Acidity")
plt.xlabel('Citric Acidity')
plt.ylabel('Count')
plt.show()
```

```python
# Create a count plot of the "Quality" feature

plt.figure(figsize=(11, 6))

sns.countplot(
    x='quality',
    data=data,
    hue='quality',
    palette='Set1',
    legend=False
)

plt.title('Count Plot of Quality')
plt.xlabel('Quality')
plt.ylabel('Count')

plt.show()
```

**Task 3 - Create a new pandas series that contains the details of the acid types for a quality**

```python
# Create a new Pandas Series called "rep_acid" that contains the details of the representative quaality for the different type of acids

rep_acid = pd.Series(index = ['fixed acidity','volatile acidity','citric acid','quality'],
                     data = [data['fixed acidity'].mean(),data['volatile acidity'].mean(),
                             data['citric acid'].mean(),data['quality'].value_counts().index[0]])
```
